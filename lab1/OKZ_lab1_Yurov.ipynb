### 1.2 Доступ к данным массива

```python
a = np.arange(1, 6)
print("a: \n", a, '\n')
print("a[1]:\n", a[1], '\n')
print("a[2:4]: \n", a[2:4], '\n')
print("a[-2:]: \n", a[-2:], '\n')
print("a[::2]: \n", a[::2], '\n')
```

```python
a = np.array([1, 2, 3, 4, 5, 6, 7, 6, 5, 4, 3, 2, 1])
b = np.copy(a)
a[a > 5] = 0
b[(b >= 3) & (b <= 5)] = 0
print("a: \n", a, '\n')
print("b: \n", b, '\n')
```

### 2.2 Простые операции с изображениями

2.2.1 Кадрирование

```python
ROI = (182, 162, 220, 240)
cropped_image = bgr_image[ROI[1]:ROI[1] + ROI[3], ROI[0]:ROI[0] + ROI[2]]
plt.imshow(cropped_image)
```

2.2.2 Изменение размера изображения

```python
scale_percent = 40
width = int(bgr_image.shape[1] * scale_percent / 100)
height = int(bgr_image.shape[0] * scale_percent / 100)
dim = (width, height)
resized_image = cv.resize(bgr_image, dim, interpolation=cv.INTER_AREA)
plt.imshow(resized_image)
```

2.2.3 Поворот изображения

```python
(h, w, d) = bgr_image.shape
angle = 45
scale = 1.0
center = (w // 2, h // 2)
M = cv.getRotationMatrix2D(center, angle, scale)
rotated = cv.warpAffine(bgr_image, M, (w, h))
plt.imshow(rotated)
```

2.2.4 Рисование примитивов на изображении

```python
copy_image = bgr_image.copy()

cv.line(copy_image, (240, 250), (360, 250), (0, 0, 0), 5)

cv.line(copy_image, (240, 250), (250, 280), (0, 0, 0), 5)
cv.line(copy_image, (250, 280), (290, 280), (0, 0, 0), 5)
cv.line(copy_image, (290, 280), (300, 250), (0, 0, 0), 5)

cv.line(copy_image, (310, 250), (320, 280), (0, 0, 0), 5)
cv.line(copy_image, (320, 280), (345, 280), (0, 0, 0), 5)
cv.line(copy_image, (345, 280), (355, 250), (0, 0, 0), 5)


plt.imshow(copy_image)
```

2.2.5 Наложение текста на изображение

```python
font = cv.FONT_HERSHEY_COMPLEX
cv.putText(copy_image, "hype.\ndrip.\nswaga.", (400, 420), font, 1, (0, 0, 0), 1, cv.LINE_AA)
plt.imshow(copy_image)
```
