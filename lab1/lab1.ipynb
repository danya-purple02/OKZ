```python
import numpy as np
import matplotlib as plt
import scipy.io as sio
```

### 1.5 Сохранение и загрузка данных на диск

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

np.savez("mas_a_b.npz", a=a, b=b)
```

```python
a_load = np.load("mas_a_b.npz")["a"]
b_load = np.load("mas_a_b.npz")["b"]

print("a_load:\n", a_load)
print("b_load:\n", b_load)
```

##### сохранение/загрузка данных в файл MatLAB (.mat)

```python
sio.savemat("mas_a_b.mat", {"a": a, "b": b})
```

```python
a_load_mat = sio.loadmat("mas_a_b.mat")["a"]
b_load_mat = sio.loadmat("mas_a_b.mat")["b"]

print("a_load_mat:\n", a_load_mat)
print("b_load_mat:\n", b_load_mat)
```

### 1.6 Приведение типов в NumPy

```python
print("a float32:\n", a.astype(np.float32))
print("a uint32:\n", a.astype(np.uint32))
```

```python
print(np.float64(11.12354))
```

```python
print(np.finfo(np.float64),"\n")
print(np.finfo(np.float32),"\n")
print(np.finfo(np.float16),"\n")
```
